```python
from tensorflow.keras.models import load_model

# Reemplaza 'nombre_del_archivo.h5' con el nombre del archivo que subiste
model = load_model('/kaggle/input/moda2/tensorflow2/default/1/fashion_model_improved.h5')


# Verificar que el modelo se cargó correctamente
model.summary()
```

```python
from tensorflow.keras.preprocessing.image import load_img, img_to_array
import numpy as np
import os

# Directorio donde están las imágenes reales
real_images_dir = '/kaggle/input/imagenes/real_images'  # Cambia esta ruta según tu entorno

# Lista de archivos de imagen (ajusta según tus imágenes reales)
image_files = ['image1.jpg', 'image2.jpg', 'image3.jpg','image4.jpg','image5.jpg','image6.jpg','image8.jpg']  # Reemplaza con los nombres de tus imágenes reales

# Lista para almacenar las imágenes cargadas
images = []

# Cargar y preprocesar las imágenes
for image_file in image_files:
    img_path = os.path.join(real_images_dir, image_file)
    img = load_img(img_path, target_size=(224, 224))  # Ajusta el tamaño según lo que el modelo espera
    img_array = img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)  # Añadir una dimensión para el lote
    img_array /= 255.0  # Normalizar la imagen si usaste rescale=1./255 durante el entrenamiento
    images.append(img_array)

# Convertir la lista de imágenes en un array numpy
images = np.vstack(images)
```

```python
# Hacer predicciones
predictions = model.predict(images)

# Convertir las predicciones a clases
predicted_classes = np.argmax(predictions, axis=1)
```
